```python
import pandas as pd
from sklearn.model_selection import train_test_split
```

```python
path = "../data/heart.csv"
df = pd.read_csv(path)
```

```python
df.info()
```

```python
df1 = df.drop_duplicates()
(df1 == df).all().all()

df_v1 = df[df['RestingBP'] > 0]
df_v1.shape
```

```python
FEATURES = [col for col in df.columns if col != "HeartDisease"]
TARGET = ['HeartDisease']

X = df_v1[FEATURES]
y = df_v1[TARGET]
```

```python
TRAIN_SIZE = 0.6
VAL_SIZE = 0.2
TEST_SIZE = VAL_SIZE / (1 - TRAIN_SIZE)

X_train, X_, y_train, y_ = train_test_split(X, y, train_size=TRAIN_SIZE, random_state=42, stratify=y)
X_val, X_test, y_val, y_test = train_test_split(X_, y_, train_size=TEST_SIZE, random_state=42, stratify=y_)
```

```python
X_train.shape[0] + X_val.shape[0] + X_test.shape[0]
```

```python
X_train.to_csv('../data/inputs_train.csv', index=False)
X_val.to_csv('../data/inputs_val.csv', index=False)
X_test.to_csv('../data/inputs_test.csv', index=False)

y_train.to_csv('../data/targets_train.csv', index=False)
y_val.to_csv('../data/targets_val.csv', index=False)
y_test.to_csv('../data/targets_test.csv', index=False)
```
